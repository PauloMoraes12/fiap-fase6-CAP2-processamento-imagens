
# Laboratório 07 — Recortes de Fotografias

## Objetivo

Aprender a selecionar regiões de uma fotografia real
utilizando o fatiamento de matrizes NumPy.

## Conteúdos

- Coordenadas de pixels
- Linhas e colunas
- Fatiamento de matrizes
- Recorte de fotografias
- Visualização e salvamento dos resultados

## Ferramentas

- Python
- NumPy
- Pillow
- Matplotlib
- Jupyter Notebook

```python

from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Carregando a fotografia
foto = Image.open("../imagens/PAULO.jpeg").convert("RGB")

# Convertendo para matriz NumPy
imagem = np.array(foto)

# Verificando as dimensões
print("Formato da imagem:", imagem.shape)
print("Tipo de dados:", imagem.dtype)

# Exibindo a fotografia
plt.figure(figsize=(8, 6))
plt.imshow(imagem)
plt.title("Fotografia original — Paulo")
plt.axis("off")
plt.show()
```

```python

# Recortando a região central da fotografia
recorte = imagem[300:1000, 300:850]

print("Formato original:", imagem.shape)
print("Formato do recorte:", recorte.shape)

plt.figure(figsize=(6, 6))
plt.imshow(recorte)
plt.title("Primeiro recorte da fotografia")
plt.axis("off")
plt.show()

print("Imagem original:", imagem.shape)
print("Recorte do rosto:", recorte.shape)
```

```python

from PIL import Image

# Convertendo a matriz NumPy em imagem
foto_recortada = Image.fromarray(recorte)

# Salvando o recorte
foto_recortada.save("../imagens/PAULO_rosto.png")

print("Recorte salvo com sucesso!")
print("Dimensões:", foto_recortada.size)
```


## Resultado — Recorte de fotografia

### Fotografia original

- Arquivo: PAULO.jpeg
- Dimensões: 1600 × 1021 pixels
- Formato NumPy: (1600, 1021, 3)

### Região selecionada

Utilizei o fatiamento de matrizes NumPy:

`recorte = imagem[300:1000, 300:850]`

- Linhas: 300 até 999
- Colunas: 300 até 849
- Altura do recorte: 700 pixels
- Largura do recorte: 550 pixels
- Formato NumPy: (700, 550, 3)

### Salvamento

Converti a matriz NumPy em imagem utilizando
`Image.fromarray()`.

O resultado foi salvo como `PAULO_rosto.png`.

### Conclusão

Aprendi a selecionar uma região específica de
uma fotografia usando índices de matrizes NumPy.

Também compreendi que o índice final do
fatiamento não é incluído e que o Pillow utiliza
a ordem (largura, altura), enquanto o NumPy
utiliza (altura, largura, canais).
