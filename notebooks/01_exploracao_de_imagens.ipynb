## Primeiro experimento

Criamos uma matriz 3 × 3 que representa intensidades
de pixels em escala de cinza.


```python
import numpy as np

imagem = np.array([
    [0, 128, 255],
    [50, 100, 200],
    [255, 128, 0]
])

print(imagem)
print("Formato:", imagem.shape)
```


## Resultado do experimento

A matriz NumPy de formato `(3, 3)` foi convertida
em uma representação visual de escala de cinza.

### Observações

- O valor `0` foi representado pela cor preta.
- O valor `128` foi representado por um cinza intermediário.
- O valor `255` foi representado pela cor branca.
- A função `imshow()` exibiu os valores como uma imagem.
- O parâmetro `cmap="gray"` definiu a escala de cinza.
- Os parâmetros `vmin=0` e `vmax=255` fixaram
  os limites de intensidade.
- A função `colorbar()` exibiu a referência das intensidades.

### Conclusão

Compreendi que uma imagem digital pode ser representada
por uma matriz numérica e que o Matplotlib permite
visualizar seus valores como intensidades de cor.

Também aprendi que o Jupyter Notebook permite executar
códigos e visualizar os resultados diretamente
abaixo das células, facilitando a análise de dados.


```python

import matplotlib.pyplot as plt

plt.figure(figsize=(5, 5))

plt.imshow(imagem, cmap="gray", vmin=0, vmax=255)

plt.title("Imagem em escala de cinza")
plt.colorbar(label="Intensidade do pixel")

plt.xticks(range(3))
plt.yticks(range(3))

plt.show()
```
