
# Laboratório 08 — Redimensionamento e Filtros

## Objetivo

Aprender a alterar as dimensões de fotografias
e aplicar filtros básicos de processamento de imagens.

## Conteúdos

- Redimensionamento de imagens
- Preservação da proporção original
- Interpolação de pixels
- Filtros de suavização e nitidez
- Comparação entre imagens processadas

## Ferramentas

- Python
- Pillow
- NumPy
- Matplotlib
- Jupyter Notebook


```python

from PIL import Image
import matplotlib.pyplot as plt

# Carregando a fotografia original
foto = Image.open("../imagens/PAULO.jpeg").convert("RGB")

# Consultando as dimensões originais
print("Dimensões originais:", foto.size)

# Redimensionando a fotografia
foto_menor = foto.resize((510, 800))

# Consultando as novas dimensões
print("Novas dimensões:", foto_menor.size)

# Comparando as fotografias
fig, eixos = plt.subplots(1, 2, figsize=(10, 7))

eixos[0].imshow(foto)
eixos[0].set_title("Original — 1021 × 1600")
eixos[0].axis("off")

eixos[1].imshow(foto_menor)
eixos[1].set_title("Redimensionada — 510 × 800")
eixos[1].axis("off")

plt.tight_layout()
plt.show()
```

```python

# Redimensionamento com dois métodos diferentes
foto_nearest = foto.resize(
    (510, 800),
    resample=Image.Resampling.NEAREST
)

foto_lanczos = foto.resize(
    (510, 800),
    resample=Image.Resampling.LANCZOS
)

# Comparando os resultados
fig, eixos = plt.subplots(1, 2, figsize=(10, 7))

eixos[0].imshow(foto_nearest)
eixos[0].set_title("Interpolação NEAREST")
eixos[0].axis("off")

eixos[1].imshow(foto_lanczos)
eixos[1].set_title("Interpolação LANCZOS")
eixos[1].axis("off")

plt.tight_layout()
plt.show()
```

```python

import numpy as np

# Convertendo as fotografias em matrizes
matriz_nearest = np.array(foto_nearest)
matriz_lanczos = np.array(foto_lanczos)

# Calculando a média dos valores RGB
media_nearest = matriz_nearest.mean()
media_lanczos = matriz_lanczos.mean()

print("Média NEAREST:", round(media_nearest, 2))
print("Média LANCZOS:", round(media_lanczos, 2))
```

```python

from PIL import ImageFilter

# Aplicando desfoque na fotografia
foto_desfocada = foto.filter(
    ImageFilter.GaussianBlur(radius=5)
)

# Comparando original e desfocada
fig, eixos = plt.subplots(1, 2, figsize=(10, 7))

eixos[0].imshow(foto)
eixos[0].set_title("Fotografia original")
eixos[0].axis("off")

eixos[1].imshow(foto_desfocada)
eixos[1].set_title("Gaussian Blur — raio 5")
eixos[1].axis("off")

plt.tight_layout()
plt.show()
```

```python

# Aplicando filtro de nitidez
foto_nitida = foto.filter(
    ImageFilter.SHARPEN
)

# Comparando as fotografias
fig, eixos = plt.subplots(1, 2, figsize=(10, 7))

eixos[0].imshow(foto)
eixos[0].set_title("Fotografia original")
eixos[0].axis("off")

eixos[1].imshow(foto_nitida)
eixos[1].set_title("Filtro SHARPEN")
eixos[1].axis("off")

plt.tight_layout()
plt.show()
```

```python

# Salvando as imagens processadas
foto_menor.save("../imagens/PAULO_redimensionada.png")

foto_desfocada.save("../imagens/PAULO_desfocada.png")

foto_nitida.save("../imagens/PAULO_nitida.png")

print("Imagens salvas com sucesso!")
```
